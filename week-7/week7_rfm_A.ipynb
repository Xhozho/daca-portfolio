```python
def load_all_rows(table_name, page_size=1000):
    all_rows = []
    start = 0

    while True:
        response = (
            supabase
            .table(table_name)
            .select("*")
            .range(start, start + page_size - 1)
            .execute()
        )

        batch = response.data
        all_rows.extend(batch)

        if len(batch) < page_size:
            break

        start += page_size

    return pd.DataFrame(all_rows)
```

```python
sales_df = load_all_rows("sales")

print("Sales shape:", sales_df.shape)
```

```python
customers_df = load_all_rows("customers")

print("Customers shape:", customers_df.shape)
```

```python
print("Sales columns:")
print(sales_df.columns.tolist())

sales_df.head()
```

```python
df = pd.merge(
    sales_df,
    customers_df,
    on="customer_id",
    how="left"
)

print("Merged shape:", df.shape)
```
